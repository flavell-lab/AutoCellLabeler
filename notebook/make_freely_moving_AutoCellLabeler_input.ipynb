# Create training data for AutoCellLabeler

This notebook takes as input the following directories:

- `path_rgb` contains RGB images for NeuroPAL aniamls containing mNeptune, OFP, and BFP images. Each image should be an `X*Y*Z*3` matrix. In the directory, each image should be saved as `dataset.nrrd`.
- `path_tagrfp` contains TagRFP (all-red) images for the same animals, saved as `dataset.nrrd`. 
- `path_roi` contains ROI images for the same animals, saved as `dataset.nrrd`.
- `path_label` contains human-labeled CSV files for the same animals, saved as `dataset Neuron ID.csv`.
- `path_neuron_ids_file` is an `h5` file that consists of all valid labels. Labels other than these are ignored in the datasets.
- `path_fm_roi_map` is an `h5` file that consists of mappings from freely moving to immobilized ROI #s.
- `path_output` contains the root path for outputs. These outputs will be the properly-formatted training, validation, and testing data for AutoCellLabeler.

The `path_rgb`, `path_tagrfp`, and `path_roi` images for a dataset respectively correspond to the `NeuroPAL.nrrd`, `all_red.nrrd`, and `neuron_rois.nrrd` files output by the ANTSUN notebook for that dataset.

It then crops the input images and formats them for input into AutoCellLabeler as training, validation, and testing data. It also crops the ROI images the same way.

The output of this notebook is also used to generate CellDiscoveryNet training data - the `make_CellDiscoveryNet_input.ipynb` notebook takes as input the output of this notebook. Notably, that notebook does not use any of the human labels as CellDiscoveryNet uses unsupervised learning. You can set `label_file=None` to generate the training data for CellDiscoveryNet.

```python
def list_files_to_dict(directory):
    """
    Helper function to list which timepoints were selected in each NRRD directory.
    """
    files = os.listdir(directory)
    uid_dict = {}
    for file in files:
        if file.endswith(".nrrd"):
            uid, timepoint = file.split("_")
            timepoint = timepoint.split(".")[0]  # Remove the extension
            if uid in uid_dict:
                uid_dict[uid].append(timepoint)
            else:
                uid_dict[uid] = [timepoint]
    return uid_dict


uid_timepoint_dict = list_files_to_dict(output_path_nrrd)

def list_files(directory):
    """
    Helper function to list which timepoints were selected in each NRRD directory.
    """
    files = os.listdir(directory)
    uid_list = []
    for file in files:
        if file.endswith(".nrrd"):
            uid_list.append(os.path.splitext(file)[0])
    return uid_list


```

### Copy files from freely-moving datasets

This block of code copies **random** time points from freely-moving datasets, directly from ANTSUN output. There are also pre-copied files available in our Dropbox under AutoCellLabeler/freely_moving_eval. If using those files, feel free to skip this section. If using ANTSUN output, it is important to run this code block.

```python
# random.seed(931)
dataset_timepoints = {}

max_t = 1600 # max timepoint in each dataset

currTimepointDir = list_files_to_dict(output_path_nrrd)

for dataset in tqdm(datasets):
    n_trials = 100
    if dataset in currTimepointDir:
        if len(currTimepointDir[dataset]) > n_trials:
            print(f"WARNING: dataset {dataset} has {len(currTimepointDir[dataset])} timepoints instead of the specified {max_t}")
            continue
        if len(currTimepointDir[dataset]) == n_trials:
            continue
        n_trials = n_trials - len(currTimepointDir[dataset])

    if dataset == "2023-06-05-17":
        prj_dir = "/data1/prj_sickness/data_processed"
    elif dataset in datasets_prj_rim:
        prj_dir = input_paths["prj_rim"]
    elif dataset in datasets_prj_neuropal:
        prj_dir = input_paths["prj_neuropal"]
    elif dataset in datasets_prj_starvation:
        prj_dir = input_paths["prj_starvation"]
    elif dataset in datasets_prj_5ht:
        prj_dir = input_paths["prj_5ht"]
    elif dataset in datasets_prj_aversion:
        prj_dir = input_paths["prj_aversion"]
    else:
        print("WARNING: Couldn't find the project of dataset " + dataset)
        continue
    
    base_path = os.path.join(prj_dir, dataset + "_output")
    dataset_timepoints[dataset] = []
    for trial in range(n_trials):
        t = np.random.randint(max_t)
        nrrd_path = os.path.join(base_path, "NRRD_cropped", dataset + "_t" + str(t).zfill(4) + "_ch2.nrrd")
        watershed_path = os.path.join(base_path, "img_roi_watershed", str(t) + ".nrrd")
        n_reattempts = 0
        while (t in dataset_timepoints[dataset] or not os.path.exists(nrrd_path) or not os.path.exists(watershed_path)) and n_reattempts < 1000:
            t = np.random.randint(max_t)
            nrrd_path = os.path.join(base_path, "NRRD_cropped", dataset + "_t" + str(t).zfill(4) + "_ch2.nrrd")
            watershed_path = os.path.join(base_path, "img_roi_watershed", str(t) + ".nrrd")
            n_reattempts += 1
        if n_reattempts >= 1000:
            print("WARNING: Could not find valid timepoint for dataset " + dataset)
            break
            # raise(ValueError("Could not find valid timepoint for dataset " + dataset))
        
        try:
            expand_nrrd_dimension(nrrd_path, os.path.join(output_path_nrrd, dataset + "_" + str(t) + ".nrrd"))
        except PermissionError:
            print(f"WARNING: Alex hasn't given you access to {os.path.join(output_path_nrrd, dataset + '_' + str(t) + '.nrrd')}")
        shutil.copy(watershed_path, os.path.join(output_path_roi, dataset + "_" + str(t) + ".nrrd"))
        dataset_timepoints[dataset].append(t)
```

### Load human labels and count rarity of labels

```python
from autolabel.roi_map import map_roi_to_neuron
from autolabel.unet_input import generate_combinations


dict_roi_to_neuron_all = {}

for dataset in tqdm(datasets):
    dict_roi_to_neuron_all[dataset] = map_roi_to_neuron(os.path.join(path_label, dataset + " Neuron ID.csv"))

extracted_neuron_ids = []
with h5py.File(path_neuron_ids_file, 'r') as f:
    extracted_neuron_ids = [name.decode('utf-8') for name in f['neuron_ids'][:]]

num_labels = {}
num_question_labels = {}

for dataset in tqdm(datasets):
    for roi in dict_roi_to_neuron_all[dataset][0].keys():
        for neuron in dict_roi_to_neuron_all[dataset][0][roi]:
            if neuron in num_labels:
                num_labels[neuron] += 1
            else:
                num_labels[neuron] = 1
            if '?' in neuron:
                for comb in generate_combinations(neuron):
                    if comb in num_question_labels:
                        num_question_labels[comb] += 1
                    else:
                        num_question_labels[comb] = 1
```

### Generate AutoCellLabeler training data

```python
from autolabel.unet_input import create_h5_from_nrrd

for dataset in tqdm(uid_timepoint_dict.keys()):
    unalignedDict = {}
    
    for t in uid_timepoint_dict[dataset]:
        timepoint = f"{dataset}_{t}"
        if dataset in datasets_test:
            save_dir = path_test
        elif dataset in datasets_val:
            save_dir = path_val
        else:
            save_dir = path_train

        # create output directories
        full_output_path = os.path.join(path_output, save_dir)
        os.makedirs(full_output_path, exist_ok=True)
        full_roi_crop_path = os.path.join(path_output, path_roi_crop)
        os.makedirs(full_roi_crop_path, exist_ok=True)
        
        output_path = os.path.join(full_output_path, timepoint + ".h5")
        roi_path = os.path.join(output_path_roi, timepoint + ".nrrd")
        roi_crop_path = os.path.join(full_roi_crop_path, timepoint + ".h5")
        label_path = os.path.join(path_label, dataset + " Neuron ID.csv")
        all_red_path = os.path.join(output_path_nrrd, timepoint + ".nrrd")

        if True and os.path.exists(roi_crop_path):
            continue

        # the `id_weight` parameter currently has little effect on the output, it describes how much smaller the weights are for wrong labels (which currently does not matter)    
        # set `label_file=None` to generate input data without labels (eg: for training CellDiscoveryNet or evaluating AutoCellLabeler on new, unlabeled datasets)
        create_h5_from_nrrd(output_path, roi_path, roi_crop_path, crop_size, num_labels, θh_pos_is_ventral[dataset], path_fm_roi_map, label_file=label_path, neuron_ids_list_file=path_neuron_ids_file, all_red_path=all_red_path, id_weight=0.3, unalignedDict=unalignedDict, freelyMoving = True)
    f = open(os.path.join(path_output, "unalignedLog.txt"), "a")
    f.write(f" - Dataset {dataset}")
    for roi, count in unalignedDict.items():
        f.write(f"Imobilized ROI {roi} ({count[1]}) not aligned {count[0]} times")
    f.close()
```
